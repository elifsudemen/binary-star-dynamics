What is a binary star system?

A star system consisting of two stars orbiting around their common center of mass.






 

Although the stars and their separations are very big, they are usually not separable in binaries with naked eyes because they are extremely far away. To resolve binaries at such great distances and beyond, scientists use various techniques. Based on these techniques, the binary stars have been classified broadly into four major categories,


1.Visual binaries
2.Spectroscopic binaries
3.Eclipsing binaries
4.Astrometric binaries



Many visual binaries have very long orbital periods, such as several centuries or millennia, and therefore their orbital periods are not known for sure or are rarely known. They can also be detected by non-direct techniques such as spectroscopy (spectroscopic binaries) or astrometry (astrometric binaries). If binary stars orbit in the same plane along our line of sight, one of their components will cover the other, that is, leave it in an eclipse, and such binaries are called a covering binary or eclipse binary.

If the components of a binary star system are close enough to each other, they can distort their gravity and switch to the opposite outer stellar atmosphere. In some cases, close binary systems undergo a process of development that single stars cannot do, and they can transfer mass. Examples of binaries are Sirius and Cygnus X-1 (known as Cygnus X-1 black hole).






Theory
Approximately half of the stars in our galaxy are members of so-called binary star systems. Such systems consist of two stars orbiting about their common center of mass. The distance separating the stars is always much less than the distance to the nearest neighbour star. Hence, a binary star system can be treated as a two-body dynamical system to a very good approximation.
In a binary star system, the gravitational force which the first star exerts on the second is

\begin{align}
f = \frac{G m_1 m_2}{r^3} r\\
\end{align}

where ${\bf r}= {\bf r}_2-{\bf r}_1$. As we have seen, a two-body system can be reduced to an equivalent one-body system whose equation of motion is of the form (327), where $\mu= m_1\,m_2/(m_1+m_2)$. Hence, in this particular case, we can write


\begin{align}
\frac{m_1m_2}{m_1+m_2}\frac{d^2\bf r}{dt^2} = - \frac{G m_1 m_2}{r^3} {\bf r}
\end{align}

which gives

$$
\begin{align}
\frac{d^2\bf r}{dt^2} = - \frac{G M}{r^3} {\bf r}
\end{align}
$$


where
$$
M = m_1 + m_2 .
$$

using the equations about the we arrived the following differential equations

$$
\ddot r- r \dot \theta^2 =-\frac{GM}{r^2}
$$
and
$$
r\ddot \theta + 2 \dot r\dot\theta =0
$$

let $$\dot r = v$$ and $$\dot \theta = \omega$$
which gives
$$
\dot v = a = r \omega^2 -\frac{GM}{r^2}
$$  which also gives $$dv=adt$$and$$dr = vdt$$
and $$
\dot\omega = -\frac{2\dot r \dot\theta}{r}
$$ which also gives $$
d\omega = \dot \omega dt 
$$ and $$
d\theta = \omega dt
$$

### Code

For these functions Numpy and Matplotlib's Pyplot class is used . 



```python
import matplotlib.pyplot as plt
import numpy as np
```

```python

def orbit(r0 = 5,v0=1,theta0=0,w0=1):

    dt = 0.05
    M= 10
    G = 0.006
    t = np.arange(0,10,dt)
    r = np.zeros(len(t))
    v = np.zeros(len(t))
    theta = np.zeros(len(t))
    w = np.zeros(len(t)) # omega: thetanin birinci zaman turevi

    r[0] = r0
    v[0] = v0
    theta[0] = theta0
    w[0] = w0 

    a = lambda r, w: (r*w**2)-(G*M/r**2)
    tt = lambda r,v,w: -2*v*w/r # thetanın ikini zaman turevi

    for i in range(len(t)-1):
        v[i+1] = v[i]+dt*a(r[i],w[i])
        r[i+1] = r[i]+dt*v[i]
        w[i+1] = w[i]+dt*tt(r[i],v[i],w[i])
        theta[i+1] = theta[i]+dt*w[i]

    plt.plot(t,r)
    plt.ylabel ('r')
    plt.xlabel ('t')
    plt.show()
    plt.plot(t,v)
    plt.ylabel ('v')
    plt.xlabel ('t')
    plt.show()
    plt.plot(t,theta)
    plt.ylabel ("theta")
    plt.xlabel ('t')
    plt.show()
    plt.plot(t,w)
    plt.ylabel ('omega')
    plt.xlabel ('t')
    plt.show()
    
    return None



```
